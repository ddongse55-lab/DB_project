res3 = enc.fit_transform(df[["Sex"]])  
<font color=blue>학습 및 변환: 원-핫 인코딩은 2차원 입력이 필요하기 때문에 [[]]으로 들어감. 1차원은 []  </font>

print(enc.categories_)  
res3.toarray()  

df3_1 = pd.DataFrame(res3.toarray(), columns=["Sex_female", "Sex_male"])   
<font color=blue>원-핫의 결과는 기본적으로 희소행렬이라 toarray()로 일반 numpy 배열로 변환DataFrame </font>
