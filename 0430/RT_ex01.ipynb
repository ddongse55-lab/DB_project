```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn import metrics
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
print('아이리스 종류 :', iris.target_names)
print('데이터 열 이름 :', iris.feature_names)
```

```python
data = pd.DataFrame(
    {
        'sepal_length':iris.data[:, 0],
        'sepal_width':iris.data[:, 1],
        'petal_length':iris.data[:, 2],
        'petal_width':iris.data[:, 3],
        'species':iris.target
    }
)
data.head(2)
```

```python
x = data[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = data['species']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=1)
x_train.shape, y_train.shape
```

### 랜덤포레스트 하이퍼 파라미터 (의사결정트리와의 차이점/다른점)
- n_estimators : 결정트리의 갯수를 지정 (default = 100)
- max_features : 최적의 분할을 위해 고려할 최대 feature 개수 (default = 'sqrt') -> 종류 : {'sqrt', 'log2', None}
- bootstrap : 부트스트랩(중복혀용 샘플링) 사용 여부 (default = True)

```python
forest = RandomForestClassifier(n_estimators=100)
forest
```

```python
forest.fit(x_train, y_train)
```

```python
y_pred = forest.predict(x_test)
print(y_pred)
print(list(y_test))
```

```python
print('정확도 :', metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

feature_importances_values = forest.feature_importances_
feature_importances = pd.Series(feature_importances_values, index = x_train.columns)
feature_top4 = feature_importances.sort_values(ascending=False)[:4]

plt.figure(figsize=(8,6))
plt.title('Top 4 Feature Importances')
sns.barplot(x=feature_top4, y=feature_top4.index)
plt.show()
```

### RandomForestRegressor

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('boston_house.csv')
df.head(2)
```

```python
x = df.drop('Target', axis=1)
y = df['Target']
```

```python
reg = RandomForestRegressor(
    n_estimators=50,
    criterion='squared_error',
    max_depth=4,
    max_features='sqrt',
    max_samples=1.0,
    bootstrap=True,
    oob_score=True,
    random_state=100
).fit(x,y)
```

```python
print(reg.oob_score_)
print('R2 :', reg.score(x, y))
```

```python
print(reg.feature_importances_)
```
