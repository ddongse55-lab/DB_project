## [문제] Seaborn이 제공하는 타이타닉 데이터에서 sex컬럼에 대한 데이터를 레이블 인코딩으로 변환하시오.

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
titanic.head(2)
```

```python
from sklearn.preprocessing import LabelEncoder

x = titanic.sex
x
```

```python
le = LabelEncoder()
le.fit(x)
sex = le.transform(x)
sex
```

```python
le.classes_
```

```python
le.inverse_transform(sex)
```

```python
titanic['sex'] = le.fit_transform(titanic['sex'])
titanic.head(2)
```
