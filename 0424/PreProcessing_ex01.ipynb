```python
import seaborn as sns

iris = sns.load_dataset('iris')
iris.head(2)
```

```python
x = iris.iloc[:, :-1]
x.head(2)
```

## scale()

```python
from sklearn.preprocessing import scale

x_scaled = scale(x)
x_scaled[:5, :]
```

```python
x_scaled.mean(axis=0)
```

```python
for scaled_mean in x_scaled.mean(axis=0):
    print('{:10.9f}'.format(scaled_mean))
```

## robust_scale()

```python
from sklearn.preprocessing import robust_scale

iris_robust_scaled = robust_scale(x)
iris_robust_scaled[:5, :]
```

## minmax_scale()

```python
from sklearn.preprocessing import minmax_scale

iris_minmax_scaled = minmax_scale(x)
iris_minmax_scaled[:5, :]
```

## maxabs_scale()

```python
from sklearn.preprocessing import maxabs_scale

iris_maxabs_scaled = maxabs_scale(x)
iris_maxabs_scaled[:5, :]
```

# StandardScaler

```python
import seaborn as sns

iris = sns.load_dataset('iris')
iris.iloc[:5, :-1]
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
iris_scaled = sc.fit_transform(iris.iloc[:, :-1])
iris_scaled[:5, :]
```

```python
iris_origin = sc.inverse_transform(iris_scaled)
iris_origin[:5, :]
```

## 레이블 인코딩

```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

iris = sns.load_dataset('iris')
y = iris.species
y
```

```python
le = LabelEncoder()
le.fit(y)
species = le.transform(y)
species
```

```python
le.classes_
```

```python
le.inverse_transform(species)
```

## 원-핫 인코딩

```python
from sklearn.preprocessing import OneHotEncoder

enc = OneHotEncoder()
enc.fit(species.reshape(-1, 1))
```

```python
iris_onehot = enc.transform((species.reshape(-1, 1)))
iris_onehot
```

#### sparse matrix(희소 행렬) : 값보다 0이 많은 행렬

```python
print(iris_onehot)
```

```python
iris_onehot.toarray()
```

```python
iris[['species']]
```

## 결측값 처리

```python
import seaborn as sns

iris = sns.load_dataset('iris')
x = iris.iloc[:, :-1]
x
```

```python
import random

for col in range(4):
    x.iloc[[random.sample(range(len(iris)), 10)], col] = float('nan')

x.head()
```

```python
x.mean(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
imp_mean.fit_transform(x)[0:5, :]
```

```python
x.median(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='median')
imp_mean.fit_transform(x)[0:5, :]
```

```python
x.mode(axis=0)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imp_mean = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
imp_mean.fit_transform(x)[0:5, :]
```
